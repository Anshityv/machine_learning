## Part A: Theoretical Questions (Q1 – Q20)


### 1. What is a Support Vector Machine (SVM)?
**Answer:**  
**Support Vector Machine (SVM)** is a powerful supervised machine learning algorithm used primarily for classification tasks, as well as regression and outlier detection. SVM works by finding an optimal hyperplane in an $N$-dimensional feature space that maximally separates data points of different classes, maximizing the margin between the closest points (support vectors) of each class.


### 2. What is the difference between Hard Margin and Soft Margin SVM?
**Answer:**  
- **Hard Margin SVM:** Assumes the training data is **strictly linearly separable** with no outliers or misclassifications allowed. It seeks a hyperplane with a maximum margin where all points satisfy $y_i(w^\cdot x_i + b) \ge 1$. It fails if data has noise or overlap.
- **Soft Margin SVM:** Introduces **slack variables ($\xi_i$)** and a regularization parameter ($C$) that permit some training points to fall within the margin or even on the wrong side of the hyperplane. This trade-off prevents overfitting and handles non-linearly separable or noisy data.


### 3. What is the mathematical intuition behind SVM?
**Answer:**  
Geometrically, the decision boundary is given by $w \cdot x + b = 0$. The distance from any point to this hyperplane is $\frac{|w \cdot x + b|}{\|w\|}$. SVM maximizes the total margin width, which is equal to $\frac{2}{\|w\|}$. Maximizing $\frac{2}{\|w\|}$ is mathematically equivalent to minimizing $\frac{1}{2}\|w\|^2$ subject to classification constraints $y_i(w \cdot x_i + b) \ge 1$ (for hard margin), transforming the problem into a convex quadratic programming optimization task.


### 4. What is the role of Lagrange Multipliers in SVM?
**Answer:**  
**Lagrange multipliers ($\alpha_i$)** are used to solve the constrained optimization problem of SVM by converting it into an unconstrained dual formulation using the Lagrangian function. This allows the algorithm to depend **only on the inner products** of data points ($x_i \cdot x_j$), enabling the application of the **kernel trick** to project data into higher-dimensional spaces without explicitly computing coordinates.


### 5. What are Support Vectors in SVM?
**Answer:**  
**Support vectors** are the specific data points that lie closest to the decision boundary (hyperplane). They are the critical elements of the dataset; if all other training points were removed, the optimal decision boundary would remain exactly the same. Their corresponding Lagrange multipliers are strictly greater than zero ($\alpha_i > 0$).


### 6. What is a Support Vector Classifier (SVC)?
**Answer:**  
**SVC** is the implementation of SVM designed for **classification problems**. It finds a decision boundary that separates data into discrete classes (binary or multi-class via OvR/OvO), maximizing the margin between classes.


### 7. What is a Support Vector Regressor (SVR)?
**Answer:**  
**SVR** is the adaptation of SVM for **regression tasks**. Instead of trying to separate classes with a maximum margin, SVR attempts to fit the error within a predefined tolerance band (called the **$\epsilon$-insensitive tube**), ensuring that deviations from the true target values are minimized within $\pm\epsilon$ while penalizing larger errors.


### 8. What is the Kernel Trick in SVM?
**Answer:**  
The **Kernel Trick** is a mathematical technique that allows SVM algorithms to solve non-linear classification problems by implicitly computing the dot products of data points in a high-dimensional (or infinite-dimensional) feature space **without ever explicitly transforming or computing the coordinates** in that space. Common kernels include Polynomial, RBF, and Sigmoid.


### 9. Compare Linear Kernel, Polynomial Kernel, and RBF Kernel.
**Answer:**  
- **Linear Kernel ($K(x, x') = x \cdot x'$):** Best for linearly separable or high-dimensional text classification; fast and simple.
- **Polynomial Kernel ($K(x, x') = (g \cdot x \cdot x' + c)^d$):** Captures feature interactions up to degree $d$; useful for image processing and complex polynomial boundaries.
- **RBF (Radial Basis Function) Kernel ($K(x, x') = \exp(-\gamma \|x - x'\|^2)$):** Maps data into an infinite-dimensional space; highly flexible, capable of handling complex non-linear and circular boundaries.


### 10. What is the effect of the C parameter in SVM?
**Answer:**  
The **$C$ parameter** controls the trade-off between achieving a wide margin and minimizing training misclassifications (regularization strength):
- **Small $C$:** Prioritizes a wider margin and smoother decision boundary, allowing more classification errors (higher bias, lower variance).
- **Large $C$:** Penalizes misclassifications heavily, forcing the model to fit training points strictly (narrower margin, lower bias, higher risk of overfitting).


### 11. What is the role of the Gamma parameter in RBF Kernel SVM?
**Answer:**  
The **$\gamma$ (gamma)** parameter defines how far the influence of a single training example reaches:
- **Low $\gamma$:** Wide radius of influence; results in smoother, more generalized decision boundaries.
- **High $\gamma$:** Narrow radius of influence; training points have localized effects, creating complex, tight, and wiggly decision boundaries that easily overfit the training data.


### 12. What is the Naïve Bayes classifier, and why is it called "Naïve"?
**Answer:**  
**Naïve Bayes** is a family of probabilistic classifiers based on Bayes' Theorem. It is called **"naïve"** because it makes the strong, simplifying assumption that all features are **conditionally independent** given the class label—an assumption rarely true in real-world data, yet the classifier performs surprisingly well in practice.


### 13. What is Bayes' Theorem?
**Answer:**  
Bayes' Theorem describes the mathematical probability of an event occurring based on prior knowledge of conditions related to the event:
$$P(y \mid x_1, x_2, \dots, x_n) = \frac{P(y) \prod_{i=1}^n P(x_i \mid y)}{P(x_1, x_2, \dots, x_n)}$$
Where $P(y|X)$ is the posterior probability, $P(y)$ is the prior, $P(x_i|y)$ is the likelihood, and $P(X)$ is the evidence.


### 14. Explain the differences between Gaussian Naïve Bayes, Multinomial Naïve Bayes, and Bernoulli Naïve Bayes.
**Answer:**  
- **Gaussian Naïve Bayes:** Assumes continuous features follow a normal (Gaussian) distribution.
- **Multinomial Naïve Bayes:** Designed for discrete frequency counts (e.g., word counts in text classification).
- **Bernoulli Naïve Bayes:** Designed for binary/boolean features (e.g., whether a word appears or does not appear in a document).


### 15. When should you use Gaussian Naïve Bayes over other variants?
**Answer:**  
Use Gaussian Naïve Bayes when your dataset consists of **continuous numerical features** that can be reasonably modeled as following a Gaussian (normal) distribution (e.g., Iris measurements, sensor readings).


### 16. What are the key assumptions made by Naïve Bayes?
**Answer:**  
1. **Conditional Independence:** All input features are conditionally independent of one another given the class label.
2. **Feature Equal Importance:** Each feature contributes equally to the final classification decision.


### 17. What are the advantages and disadvantages of Naïve Bayes?
**Answer:**  
- **Advantages:** Extremely fast training and prediction, highly scalable to large datasets and high-dimensional spaces (like text), requires little training data, and is robust to irrelevant features.
- **Disadvantages:** The conditional independence assumption is rarely met in reality, and it suffers from zero-frequency probability issues if a category/word is unseen in training.


### 18. Why is Naïve Bayes a good choice for text classification?
**Answer:**  
Text datasets feature extremely high dimensionality (vocabulary size) and sparse matrices. Multinomial and Bernoulli Naïve Bayes handle high feature counts efficiently, require minimal memory, train rapidly, and handle word-frequency counting natively with exceptional accuracy (e.g., spam filtering, sentiment analysis).


### 19. Compare SVM and Naïve Bayes for classification tasks.
**Answer:**  
| Aspect | SVM | Naïve Bayes |
|---|---|---|
| **Model Type** | Geometric / Maximum Margin | Probabilistic / Bayesian |
| **Independence Assumption** | Does not assume feature independence | Assumes strict conditional feature independence |
| **Speed & Scalability** | Slower training on large datasets ($O(n^2)$ to $O(n^3)$) | Extremely fast and scalable ($O(n)$) |
| **Performance** | High accuracy on complex, non-linear boundaries | Excellent on text classification and high-sparse data |


### 20. How does Laplace Smoothing help in Naïve Bayes?
**Answer:**  
**Laplace Smoothing (add-alpha smoothing)** adds a small constant ($\alpha$, typically $1$) to feature counts in the numerator and the number of feature categories in the denominator. This prevents zero probabilities ($P(x_i|y) = 0$) from wiping out the entire posterior probability calculation when an unseen feature value appears in test data.

## Part B: Practical Questions (Q21 – Q46)

Run the setup cell below first to import required libraries and datasets so all practical exercises run cleanly in Google Colab.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris, load_wine, load_breast_cancer, fetch_california_housing, fetch_20newsgroups
from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold, cross_val_score
from sklearn.svm import SVC, SVR
from sklearn.naive_bayes import GaussianNB, MultinomialNB, BernoulliNB
from sklearn.preprocessing import StandardScaler
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_selection import RFE
from sklearn.metrics import (
    accuracy_score, mean_squared_error, mean_absolute_error,
    precision_score, recall_score, f1_score, log_loss,
    roc_auc_score, confusion_matrix, ConfusionMatrixDisplay,
    PrecisionRecallDisplay
)

sns.set_theme(style="whitegrid")
print("Environment initialized successfully!")
```

### 21. Write a Python program to train an SVM Classifier on the Iris dataset and evaluate accuracy.


```python
iris = load_iris()
Xi_train, Xi_test, yi_train, yi_test = train_test_split(iris.data, iris.target, test_size=0.2, random_state=42)
scaler = StandardScaler()
Xi_tr_s = scaler.fit_transform(Xi_train)
Xi_te_s = scaler.transform(Xi_test)

svm_iris = SVC(kernel='rbf', random_state=42).fit(Xi_tr_s, yi_train)
print(f"Q21 - SVM Classifier Accuracy (Iris): {accuracy_score(yi_test, svm_iris.predict(Xi_te_s)):.4f}")
```

### 22. Write a Python program to train two SVM classifiers with Linear and RBF kernels on the Wine dataset, then compare their accuracies.


```python
wine = load_wine()
Xw_train, Xw_test, yw_train, yw_test = train_test_split(wine.data, wine.target, test_size=0.25, random_state=42)
Xw_tr_s = scaler.fit_transform(Xw_train)
Xw_te_s = scaler.transform(Xw_test)

for kern in ['linear', 'rbf']:
    clf_k = SVC(kernel=kern, random_state=42).fit(Xw_tr_s, yw_train)
    acc = accuracy_score(yw_test, clf_k.predict(Xw_te_s))
    print(f"Q22 - Kernel: {kern:6s} -> Wine Accuracy: {acc:.4f}")
```

### 23. Write a Python program to train an SVM Regressor (SVR) on a housing dataset and evaluate it using Mean Squared Error (MSE).


```python
cal = fetch_california_housing()
Xh_train, Xh_test, yh_train, yh_test = train_test_split(cal.data, cal.target, test_size=0.2, random_state=42)
Xh_tr_s = scaler.fit_transform(Xh_train)
Xh_te_s = scaler.transform(Xh_test)

svr = SVR(kernel='rbf').fit(Xh_tr_s, yh_train)
print(f"Q23 - SVR Mean Squared Error (California Housing): {mean_squared_error(yh_test, svr.predict(Xh_te_s)):.4f}")
```

### 24. Write a Python program to train an SVM Classifier with a Polynomial Kernel and visualize the decision boundary.


```python
X24 = iris.data[:, :2]
y24 = iris.target
svm_poly = SVC(kernel='poly', degree=3, C=1.0, random_state=42).fit(X24, y24)

x_min, x_max = X24[:, 0].min() - 0.5, X24[:, 0].max() + 0.5
y_min, y_max = X24[:, 1].min() - 0.5, X24[:, 1].max() + 0.5
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.02), np.arange(y_min, y_max, 0.02))
Z24 = svm_poly.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

plt.figure(figsize=(7, 5))
plt.contourf(xx, yy, Z24, alpha=0.3, cmap='viridis')
plt.scatter(X24[:, 0], X24[:, 1], c=y24, cmap='viridis', edgecolor='k', s=45)
plt.title("Q24: SVM Polynomial Kernel Decision Boundary (Iris)")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.show()
```
