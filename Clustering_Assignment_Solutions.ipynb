## Part A: Theoretical Questions (Q1 – Q20)


### 1. What is unsupervised learning in the context of machine learning?
**Answer:**  
**Unsupervised learning** is a machine learning paradigm where algorithms learn patterns, structures, or groupings directly from **unlabeled input data** ($X$) without explicit target outputs ($Y$). Common tasks include clustering (e.g., K-Means, DBSCAN), dimensionality reduction (e.g., PCA, t-SNE), and association rule mining.


### 2. How does K-Means clustering algorithm work?
**Answer:**  
1. **Initialization:** Choose the number of clusters $K$ and initialize $K$ centroids (randomly or via K-Means++).
2. **Assignment Step:** Assign every data point to its nearest centroid based on Euclidean distance.
3. **Update Step:** Recompute each centroid as the arithmetic mean of all data points assigned to that cluster.
4. **Convergence:** Repeat Steps 2 and 3 iteratively until centroid positions stabilize or maximum iterations are reached.


### 3. Explain the concept of a dendrogram in hierarchical clustering.
**Answer:**  
A **dendrogram** is a tree-like diagram that illustrates the hierarchical sequence of merges (in agglomerative clustering) or splits (in divisive clustering). The horizontal axis lists data points/clusters, while the vertical axis represents the Euclidean distance or dissimilarity at which clusters are merged. Cutting the dendrogram horizontally at a chosen distance height yields a specific number of flat clusters.


### 4. What is the main difference between K-Means and Hierarchical Clustering?
**Answer:**  
- **Cluster Count:** K-Means requires pre-specifying the exact number of clusters $K$ beforehand, whereas Hierarchical Clustering builds a complete hierarchy (dendrogram) without needing $K$ in advance.
- **Scalability & Shape:** K-Means scales linearly ($O(n)$) to large datasets and assumes spherical clusters, while Hierarchical Clustering has higher complexity ($O(n^2)$ to $O(n^3)$) and captures nested cluster structures.


### 5. What are the advantages of DBSCAN over K-Means?
**Answer:**  
- Does **not** require pre-specifying the number of clusters $K$.
- Can discover **arbitrarily shaped, non-linear clusters** (e.g., concentric circles or moons), whereas K-Means only finds convex/spherical clusters.
- Automatically identifies and isolates **outliers/noise points** (labeled `-1`), making it robust against extreme values.


### 6. When would you use Silhouette Score in clustering?
**Answer:**  
Silhouette Score is used when ground-truth labels are unavailable to evaluate **cluster cohesion** (how close points are within the same cluster) versus **cluster separation** (how far apart distinct clusters are). It is especially helpful for selecting the optimal number of clusters $K$ and comparing clustering algorithms on the same dataset.


### 7. What are the limitations of Hierarchical Clustering?
**Answer:**  
- **High Time & Memory Complexity:** Requires computing and storing a pairwise distance matrix ($O(n^2)$ space and $O(n^2 \log n)$ or $O(n^3)$ time), making it impractical for large datasets.
- **Irreversible Decisions:** Once two clusters are merged or split, the step cannot be undone later.
- **Sensitivity to Noise & Outliers:** Outliers can distort linkage distances (especially in single or complete linkage).


### 8. Why is feature scaling important in clustering algorithms like K-Means?
**Answer:**  
K-Means relies on **Euclidean distance** to assign points to centroids. If features are on vastly different scales (e.g., annual income in lakhs vs. age in years), features with larger numeric ranges will completely dominate distance calculations. Standardizing (`StandardScaler`) or normalizing (`MinMaxScaler`) ensures all features contribute equally.


### 9. How does DBSCAN identify noise points?
**Answer:**  
DBSCAN uses two parameters: neighborhood radius `eps` ($\epsilon$) and minimum points `min_samples`. A point is a **core point** if at least `min_samples` fall within distance $\epsilon$; a **border point** lies within $\epsilon$ of a core point but has fewer than `min_samples` neighbors; any point that is neither a core point nor reachable from a core point is labeled as **noise (`-1`)**.


### 10. Define inertia in the context of K-Means.
**Answer:**  
**Inertia** (Within-Cluster Sum of Squares, WCSS) measures how internally coherent clusters are by summing the squared Euclidean distances between every data point and its assigned cluster centroid:
$$\text{Inertia} = \sum_{i=1}^{N} \|x_i - \mu_{c_i}\|^2$$
Lower inertia indicates tighter clusters, though inertia always decreases as $K$ increases.


### 11. What is the elbow method in K-Means clustering?
**Answer:**  
The **Elbow Method** is a heuristic for choosing the optimal $K$. K-Means is run across a range of $K$ values (e.g., $K=1$ to $10$) and inertia is plotted against $K$. The point where the sharp rate of decrease in inertia bends and flattens out—resembling an "elbow"—indicates the optimal trade-off between compact clusters and model parsimony.


### 12. Describe the concept of "density" in DBSCAN.
**Answer:**  
In DBSCAN, **density** is defined as the number of data points residing within a hypersphere of radius $\epsilon$ (`eps`) centered around a given point. Regions where point count meets or exceeds `min_samples` are treated as dense clusters, separated by low-density regions.


### 13. Can hierarchical clustering be used on categorical data?
**Answer:**  
**Yes**, provided an appropriate categorical distance/dissimilarity metric is used instead of Euclidean distance—such as **Hamming distance**, **Jaccard distance**, or **Gower distance** (for mixed numeric and categorical data)—alongside a precomputed distance matrix.


### 14. What does a negative Silhouette Score indicate?
**Answer:**  
A **negative Silhouette Score** (closer to $-1$) indicates that data points have likely been assigned to the **wrong cluster**, as their average distance to points in a neighboring cluster is smaller than their average distance to points within their own assigned cluster.


### 15. Explain the term "linkage criteria" in hierarchical clustering.
**Answer:**  
**Linkage criteria** define how the distance between two clusters is calculated during hierarchical merging:
- **Single Linkage:** Minimum distance between any two points in different clusters.
- **Complete Linkage:** Maximum distance between any two points in different clusters.
- **Average Linkage:** Mean pairwise distance between all points in the two clusters.
- **Ward Linkage:** Minimizes the total within-cluster variance increase when merging two clusters.


### 16. Why might K-Means clustering perform poorly on data with varying cluster sizes or densities?
**Answer:**  
K-Means partitions space into equal-radius Voronoi cells around centroids and minimizes total squared Euclidean distance. Consequently, it tends to split large or sparse clusters into pieces while merging small, dense clusters together.


### 17. What are the core parameters in DBSCAN, and how do they influence clustering?
**Answer:**  
1. **`eps` ($\epsilon$):** Maximum radius of the neighborhood around a point. Too small $\rightarrow$ most points become noise; too large $\rightarrow$ distinct clusters merge together.
2. **`min_samples`:** Minimum number of points within `eps` required to form a dense region (core point). Higher values create stricter density thresholds, classifying more sparse points as noise.


### 18. How does K-Means++ improve upon standard K-Means initialization?
**Answer:**  
Standard random initialization can place centroids too close together, causing slow convergence or poor local minima. **K-Means++** selects the first centroid uniformly at random and chooses subsequent centroids with probability proportional to their **squared distance from the nearest existing centroid**, spreading initial centroids evenly across the data.


### 19. What is agglomerative clustering?
**Answer:**  
**Agglomerative clustering** is a "bottom-up" hierarchical clustering approach. It starts by treating each individual data point as its own singleton cluster and iteratively merges the closest pair of clusters based on a linkage criterion until all points belong to a single cluster (or a target `n_clusters` is reached).


### 20. What makes Silhouette Score a better metric than just inertia for model evaluation?
**Answer:**  
Inertia monotonically decreases toward $0$ as $K$ approaches $N$ (offering no natural penalty for over-segmenting data) and ignores inter-cluster separation. **Silhouette Score** is bounded between $[-1, 1]$ and simultaneously evaluates **both intra-cluster cohesion and inter-cluster separation**, making it scale-invariant and reliable for comparing different $K$ values.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import make_blobs, make_moons, make_circles, load_iris, load_wine, load_breast_cancer, load_digits
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.metrics import silhouette_score, silhouette_samples
from scipy.cluster.hierarchy import dendrogram, linkage

sns.set_theme(style="whitegrid")
print("Libraries loaded successfully!")
```

### 21. Generate synthetic data with 4 centers using `make_blobs` and apply K-Means clustering. Visualize using a scatter plot.

### 44. Load the Iris dataset and plot the Silhouette Coefficient for each sample after KMeans clustering.


```python
km44 = KMeans(n_clusters=3, random_state=42, n_init=10)
labels44 = km44.fit_predict(iris.data)
sample_sil_vals = silhouette_samples(iris.data, labels44)
avg_sil = silhouette_score(iris.data, labels44)

plt.figure(figsize=(8, 5))
y_lower = 10
for i in range(3):
    ith_cluster_vals = np.sort(sample_sil_vals[labels44 == i])
    size_i = ith_cluster_vals.shape[0]
    y_upper = y_lower + size_i
    plt.fill_betweenx(np.arange(y_lower, y_upper), 0, ith_cluster_vals, alpha=0.8, label=f'Cluster {i}')
    y_lower = y_upper + 10

plt.axvline(x=avg_sil, color='red', linestyle='--', label=f'Avg Silhouette ({avg_sil:.2f})')
plt.title("Q44: Sample-wise Silhouette Plot for Iris K-Means Clustering")
plt.xlabel("Silhouette Coefficient")
plt.ylabel("Cluster Samples")
plt.legend()
plt.show()
```
