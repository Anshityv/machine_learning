### Question 1: What is Simple Linear Regression (SLR)? Explain its purpose.
**Answer:**  
**Simple Linear Regression (SLR)** is a parametric supervised machine learning algorithm used to model the linear relationship between a single independent predictor variable ($X$) and a continuous dependent target variable ($Y$).

**Purpose:**  
1. **Predictive Modeling:** To forecast future or unknown continuous target values based on a known feature input.
2. **Relationship Analysis:** To quantify the strength, direction, and statistical significance of the relationship between two continuous variables.
3. **Decision Making:** To provide actionable insights into how manipulating an input variable impacts an output metric.

### Question 2: What are the key assumptions of Simple Linear Regression?
**Answer:**  
1. **Linearity:** The relationship between the independent variable $X$ and dependent variable $Y$ is linear.
2. **Independence of Errors:** Residual errors are independent of one another (no autocorrelation/serial correlation).
3. **Homoscedasticity:** The variance of the error terms is constant across all levels of the predictor $X$.
4. **Normality of Residuals:** Residual errors follow a normal distribution with a mean of zero ($\mathcal{N}(0, \sigma^2)$).
5. **Exogeneity:** The expected value of the error term given $X$ is zero ($E(\epsilon|X) = 0$).

### Question 3: Write the mathematical equation for a simple linear regression model and explain each term.
**Answer:**  
The mathematical equation for a simple linear regression model is:
$$Y = mX + c + \epsilon$$
(Alternatively written as: $Y = \beta_0 + \beta_1 X + \epsilon$)

**Explanation of Terms:**  
- **$Y$ (Dependent Variable):** The target variable or outcome we want to predict or explain.
- **$X$ (Independent Variable):** The predictor or feature used to predict $Y$.
- **$m$ or $\beta_1$ (Slope):** Represents the regression coefficient; the expected change in $Y$ per 1-unit increase in $X$.
- **$c$ or $\beta_0$ ($Y$-Intercept):** The expected value of $Y$ when the independent variable $X = 0$.
- **$\epsilon$ (Error Term / Residual):** Accounts for random variation or unobserved factors influencing $Y$ that are not captured by feature $X$.

### Question 5: What is the method of least squares in linear regression?
**Answer:**  
The **Method of Least Squares** (Ordinary Least Squares - OLS) is a mathematical optimization technique used to determine the best-fitting line by minimizing the **Sum of Squared Residuals (SSR)** between the actual observed target values ($Y_i$) and the predicted values ($\hat{Y}_i$).

$$\text{SSR} = \sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2 = \sum_{i=1}^{n} (Y_i - (mX_i + c))^2$$

By squaring the errors, large deviations are penalized heavily, and calculus partial derivatives are set to zero to derive exact closed-form formulas for slope ($m$) and intercept ($c$).

### Question 6: What is Logistic Regression? How does it differ from Linear Regression?
**Answer:**  
**Logistic Regression** is a classification algorithm used to predict the probability of a categorical dependent outcome (typically binary, e.g., 0 or 1, Yes or No) using independent predictor variables. It applies the sigmoid function to map predictions between 0 and 1.

**Key Differences:**
| Feature | Linear Regression | Logistic Regression |
|---|---|---|
| **Task Type** | Regression (Continuous target) | Classification (Discrete/Categorical target) |
| **Output Range** | $(-\infty, \infty)$ | $[0, 1]$ (Probability values) |
| **Activation Function** | Identity function ($y = x$) | Sigmoid / Logistic function ($p = \frac{1}{1 + e^{-z}}$) |
| **Optimization Loss** | Ordinary Least Squares (OLS) | Maximum Likelihood Estimation (MLE) / Log-Loss |

### Question 7: Name and briefly describe three common evaluation metrics for regression models.
**Answer:**  
1. **Mean Absolute Error (MAE):**  
   $$\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} |Y_i - \hat{Y}_i|$$
   Measures the average magnitude of absolute errors between predictions and actual values. Highly interpretable as it shares the exact units of target variable $Y$.

2. **Mean Squared Error (MSE):**  
   $$\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2$$
   Computes the average squared difference between estimates and actual values. Squares errors, meaning larger outliers are heavily penalized.

3. **Root Mean Squared Error (RMSE):**  
   $$\text{RMSE} = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2}$|
   The square root of MSE. It brings the error metric back into the original units of $Y$, making it intuitive while retaining heavier penalties for large errors.

### Question 8: What is the purpose of the R-squared metric in regression analysis?
**Answer:**  
The **$R^2$ (Coefficient of Determination)** measures the goodness of fit of a regression model. Specifically, it represents the proportion of variance in the dependent variable $Y$ that is predictably explained by the independent predictor variable(s) $X$.

$$R^2 = 1 - \frac{\sum (Y_i - \hat{Y}_i)^2}{\sum (Y_i - \bar{Y})^2}$$

- $R^2 = 1.0$: Perfect predictive fit.
- $R^2 = 0.0$: Model performs no better than a horizontal mean baseline.
- Negative values: Model performs worse than a simple horizontal average.

### Question 9: Write Python code to fit a simple linear regression model using scikit-learn and print the slope and intercept.
*(Include your Python code and output in the code box below.)*

```python
# Import required libraries
import numpy as np
from sklearn.linear_model import LinearRegression

# Sample Data: Independent variable X (e.g., Experience in years), Dependent variable Y (Salary in LPA)
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([3.5, 5.0, 6.2, 7.8, 9.0])

# Initialize and fit the Simple Linear Regression model
model = LinearRegression()
model.fit(X, y)

# Extract and print slope (coefficient) and intercept
slope = model.coef_[0]
intercept = model.intercept_

print(f'Estimated Slope (m): {slope:.4f}')
print(f'Estimated Intercept (c): {intercept:.4f}')

# Prediction test
y_pred = model.predict(X)
print(f'Predictions: {y_pred}')
```

### Question 10: How do you interpret the coefficients in a simple linear regression model?
**Answer:**  
In a simple linear regression model ($Y = mX + c$):
1. **Slope ($m$):** Interpreted as the expected change (increase if positive, decrease if negative) in the dependent variable $Y$ for every exact **1-unit increase** in the independent variable $X$, assuming all other unstated external conditions remain constant.
2. **Intercept ($c$):** Interpreted as the baseline estimated value of $Y$ when the independent predictor $X$ is exactly equal to zero ($0$). (Note: Meaningful only if $X=0$ falls within a valid, realistic range of the data distribution).
